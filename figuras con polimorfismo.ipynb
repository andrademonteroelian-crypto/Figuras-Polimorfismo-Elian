```python
"""figuras con polimorfismo.ipynb
Automatically generated by Colab.
Original file is located at
   https://colab.research.google.com/drive/1prXRLzbyyh6rApysWN-Pbeumcin98C3q
"""
import math


class FigurasGeometricas:
    """Clase base que representa una figura geométrica con encapsulamiento."""

    def __init__(self, alto, ancho):
        # Atributos privados (encapsulamiento con doble guion bajo)
        self.__alto = alto
        self.__ancho = ancho

    # ── PROPERTY y SETTER para alto ──────────────────────────────────────────

    @property
    def alto(self):
        """Getter: retorna el valor del alto."""
        return self.__alto

    @alto.setter
    def alto(self, valor):
        """Setter: valida y asigna el alto."""
        if valor <= 0:
            raise ValueError("El alto debe ser un valor positivo.")
        self.__alto = valor

    # ── PROPERTY y SETTER para ancho ─────────────────────────────────────────

    @property
    def ancho(self):
        """Getter: retorna el valor del ancho."""
        return self.__ancho

    @ancho.setter
    def ancho(self, valor):
        """Setter: valida y asigna el ancho."""
        if valor <= 0:
            raise ValueError("El ancho debe ser un valor positivo.")
        self.__ancho = valor

    # ── Métodos sobrescritos en subclases ─────────────────────────────────────

    def area(self):
        """Calcula el área (método a sobrescribir)."""
        return self.__alto * self.__ancho

    def perimetro(self):
        """Calcula el perímetro (método a sobrescribir)."""
        return 2 * (self.__alto + self.__ancho)

    def __str__(self):
        return (f"Figura Geométrica | Alto: {self.__alto} cm | "
                f"Ancho: {self.__ancho} cm")


# ── Subclase: Rectángulo ──────────────────────────────────────────────────────

class Rectangulo(FigurasGeometricas):
    """Hereda de FigurasGeometricas. Representa un rectángulo."""

    def __init__(self, alto, ancho):
        super().__init__(alto, ancho)

    def area(self):
        """Sobrescritura: área del rectángulo."""
        return self.alto * self.ancho

    def perimetro(self):
        """Sobrescritura: perímetro del rectángulo."""
        return 2 * (self.alto + self.ancho)

    def __str__(self):
        return (f"Rectángulo       | Alto: {self.alto} cm | "
                f"Ancho: {self.ancho} cm | "
                f"Área: {self.area():.2f} cm² | "
                f"Perímetro: {self.perimetro():.2f} cm")


# ── Subclase: Triángulo ───────────────────────────────────────────────────────

class Triangulo(FigurasGeometricas):
    """Hereda de FigurasGeometricas. Representa un triángulo (base × altura)."""

    def __init__(self, alto, ancho):
        super().__init__(alto, ancho)

    def area(self):
        """Sobrescritura: área del triángulo = (base × altura) / 2."""
        return (self.ancho * self.alto) / 2

    def perimetro(self):
        """Sobrescritura: hipotenusa aproximada usando Pitágoras."""
        hipotenusa = math.sqrt(self.alto**2 + self.ancho**2)
        return self.alto + self.ancho + hipotenusa

    def __str__(self):
        return (f"Triángulo        | Alto: {self.alto} cm | "
                f"Base: {self.ancho} cm | "
                f"Área: {self.area():.2f} cm² | "
                f"Perímetro: {self.perimetro():.2f} cm")


# ── Subclase: Elipse ──────────────────────────────────────────────────────────

class Elipse(FigurasGeometricas):
    """Hereda de FigurasGeometricas. Representa una elipse (semiejes a y b)."""

    def __init__(self, alto, ancho):
        super().__init__(alto, ancho)

    def area(self):
        """Sobrescritura: área de la elipse = π × a × b."""
        return math.pi * self.alto * self.ancho

    def perimetro(self):
        """Sobrescritura: aproximación de Ramanujan para el perímetro."""
        a, b = self.alto, self.ancho
        h = ((a - b) ** 2) / ((a + b) ** 2)
        return math.pi * (a + b) * (1 + (3 * h) / (10 + math.sqrt(4 - 3 * h)))

    def __str__(self):
        return (f"Elipse           | Semi-eje a: {self.alto} cm | "
                f"Semi-eje b: {self.ancho} cm | "
                f"Área: {self.area():.2f} cm² | "
                f"Perímetro: {self.perimetro():.2f} cm")


# ── PROGRAMA PRINCIPAL ────────────────────────────────────────────────────────

if __name__ == "__main__":
    separador = "=" * 65

    print(separador)
    print("       Taller en Clase (POO en Python)")
    print(separador)

    # --- Figura base ---
    figura = FigurasGeometricas(3, 7)
    print("\n[1] Figura base creada:")
    print(f"    alto  (getter) → {figura.alto} cm")
    print(f"    ancho (getter) → {figura.ancho} cm")

    # Usando setters para modificar valores
    figura.alto = 11
    figura.ancho = 20
    print("\n[2] Valores modificados con setters:")
    print(f"    alto  (setter) → {figura.alto} cm")
    print(f"    ancho (setter) → {figura.ancho} cm")
    print(f"    Área base      → {figura.area():.2f} cm²")
    print(f"    Perímetro base → {figura.perimetro():.2f} cm")

    # Validación del setter
    print("\n[3] Validación del setter (valor negativo):")
    try:
        figura.alto = -5
    except ValueError as e:
        print(f"     Error capturado correctamente: {e}")

    # --- Herencia y sobrescritura ---
    print(f"\n{separador}")
    print("       HERENCIA Y SOBRESCRITURA DE MÉTODOS")
    print(separador)

    rect = Rectangulo(8, 15)
    tri  = Triangulo(6, 5)
    eli  = Elipse(10, 4)

    print(f"\n  {rect}")
    print(f"  {tri}")
    print(f"  {eli}")

    # Modificar con setter desde subclase
    print(f"\n[4] Cambiar dimensiones del rectángulo con setter:")
    rect.alto  = 14
    rect.ancho = 9
    print(f"  {rect}")

    print(f"\n{separador}")
    print("  Código ejecutado correctamente   ")
    print(separador)

import math


class FigurasGeometricas:
    """Clase base que representa una figura geométrica con encapsulamiento."""

    # Colores ANSI para la terminal
    RESET  = "\033[0m"
    BOLD   = "\033[1m"

    # Color base (blanco/neutro)
    COLOR  = "\033[97m"

    def __init__(self, alto, ancho):
        # Atributos privados (encapsulamiento con doble guion bajo)
        self.__alto = alto
        self.__ancho = ancho

    # ── PROPERTY y SETTER para alto ──────────────────────────────────────────

    @property
    def alto(self):
        """Getter: retorna el valor del alto."""
        return self.__alto

    @alto.setter
    def alto(self, valor):
        """Setter: valida y asigna el alto."""
        if valor <= 0:
            raise ValueError("El alto debe ser un valor positivo.")
        self.__alto = valor

    # ── PROPERTY y SETTER para ancho ─────────────────────────────────────────

    @property
    def ancho(self):
        """Getter: retorna el valor del ancho."""
        return self.__ancho

    @ancho.setter
    def ancho(self, valor):
        """Setter: valida y asigna el ancho."""
        if valor <= 0:
            raise ValueError("El ancho debe ser un valor positivo.")
        self.__ancho = valor

    # ── Métodos sobrescritos en subclases ─────────────────────────────────────

    def area(self):
        """Calcula el área (método a sobrescribir)."""
        return self.__alto * self.__ancho

    def perimetro(self):
        """Calcula el perímetro (método a sobrescribir)."""
        return 2 * (self.__alto + self.__ancho)

    def describir(self):
        """Método polimórfico: describe la figura (a sobrescribir)."""
        return "Soy una figura geométrica genérica."

    def __str__(self):
        c = self.__class__.COLOR
        r = self.RESET
        return (f"{c}Figura Geométrica | Alto: {self.__alto} cm | "
                f"Ancho: {self.__ancho} cm{r}")


# ── Subclase: Rectángulo ──────────────────────────────────────────────────────

class Rectangulo(FigurasGeometricas):
    """Hereda de FigurasGeometricas. Representa un rectángulo."""

    COLOR = "\033[94m"   # Azul

    def __init__(self, alto, ancho):
        super().__init__(alto, ancho)

    def area(self):
        """Sobrescritura: área del rectángulo."""
        return self.alto * self.ancho

    def perimetro(self):
        """Sobrescritura: perímetro del rectángulo."""
        return 2 * (self.alto + self.ancho)

    def describir(self):
        """Método polimórfico: descripción específica del rectángulo."""
        return (f"Soy un Rectángulo de {self.alto} cm de alto "
                f"y {self.ancho} cm de ancho.")

    def __str__(self):
        c = self.__class__.COLOR
        r = self.RESET
        return (f"{c}Rectángulo       | Alto: {self.alto} cm | "
                f"Ancho: {self.ancho} cm | "
                f"Área: {self.area():.2f} cm² | "
                f"Perímetro: {self.perimetro():.2f} cm{r}")


# ── Subclase: Triángulo ───────────────────────────────────────────────────────

class Triangulo(FigurasGeometricas):
    """Hereda de FigurasGeometricas. Representa un triángulo (base × altura)."""

    COLOR = "\033[92m"   # Verde

    def __init__(self, alto, ancho):
        super().__init__(alto, ancho)

    def area(self):
        """Sobrescritura: área del triángulo = (base × altura) / 2."""
        return (self.ancho * self.alto) / 2

    def perimetro(self):
        """Sobrescritura: hipotenusa aproximada usando Pitágoras."""
        hipotenusa = math.sqrt(self.alto**2 + self.ancho**2)
        return self.alto + self.ancho + hipotenusa

    def describir(self):
        """Método polimórfico: descripción específica del triángulo."""
        return (f"Soy un Triángulo con altura {self.alto} cm "
                f"y base {self.ancho} cm.")

    def __str__(self):
        c = self.__class__.COLOR
        r = self.RESET
        return (f"{c}Triángulo        | Alto: {self.alto} cm | "
                f"Base: {self.ancho} cm | "
                f"Área: {self.area():.2f} cm² | "
                f"Perímetro: {self.perimetro():.2f} cm{r}")


# ── Subclase: Elipse ──────────────────────────────────────────────────────────

class Elipse(FigurasGeometricas):
    """Hereda de FigurasGeometricas. Representa una elipse (semiejes a y b)."""

    COLOR = "\033[93m"   # Amarillo

    def __init__(self, alto, ancho):
        super().__init__(alto, ancho)

    def area(self):
        """Sobrescritura: área de la elipse = π × a × b."""
        return math.pi * self.alto * self.ancho

    def perimetro(self):
        """Sobrescritura: aproximación de Ramanujan para el perímetro."""
        a, b = self.alto, self.ancho
        h = ((a - b) ** 2) / ((a + b) ** 2)
        return math.pi * (a + b) * (1 + (3 * h) / (10 + math.sqrt(4 - 3 * h)))

    def describir(self):
        """Método polimórfico: descripción específica de la elipse."""
        return (f"Soy una Elipse con semi-eje a={self.alto} cm "
                f"y semi-eje b={self.ancho} cm.")

    def __str__(self):
        c = self.__class__.COLOR
        r = self.RESET
        return (f"{c}Elipse           | Semi-eje a: {self.alto} cm | "
                f"Semi-eje b: {self.ancho} cm | "
                f"Área: {self.area():.2f} cm² | "
                f"Perímetro: {self.perimetro():.2f} cm{r}")


# ── PROGRAMA PRINCIPAL ────────────────────────────────────────────────────────

if __name__ == "__main__":
    CYAN    = "\033[96m"
    MAGENTA = "\033[95m"
    RESET   = "\033[0m"
    BOLD    = "\033[1m"

    separador = BOLD + CYAN + "=" * 65 + RESET

    print(separador)
    print(BOLD + CYAN + "       Taller en Clase (POO en Python)" + RESET)
    print(separador)

    # --- Figura base ---
    figura = FigurasGeometricas(3, 7)
    print("\n[1] Figura base creada:")
    print(f"    alto  (getter) → {figura.alto} cm")
    print(f"    ancho (getter) → {figura.ancho} cm")

    # Usando setters para modificar valores
    figura.alto = 11
    figura.ancho = 20
    print("\n[2] Valores modificados con setters:")
    print(f"    alto  (setter) → {figura.alto} cm")
    print(f"    ancho (setter) → {figura.ancho} cm")
    print(f"    Área base      → {figura.area():.2f} cm²")
    print(f"    Perímetro base → {figura.perimetro():.2f} cm")

    # Validación del setter
    print("\n[3] Validación del setter (valor negativo):")
    try:
        figura.alto = -5
    except ValueError as e:
        print(f"     Error capturado correctamente: {e}")

    # --- Herencia y sobrescritura ---
    print(f"\n{separador}")
    print(BOLD + CYAN + "       HERENCIA Y SOBRESCRITURA DE MÉTODOS" + RESET)
    print(separador)

    rect = Rectangulo(8, 15)
    tri  = Triangulo(6, 5)
    eli  = Elipse(10, 4)

    print(f"\n  {rect}")
    print(f"  {tri}")
    print(f"  {eli}")

    # Modificar con setter desde subclase
    print(f"\n[4] Cambiar dimensiones del rectángulo con setter:")
    rect.alto  = 14
    rect.ancho = 9
    print(f"  {rect}")

    # ── POLIMORFISMO ──────────────────────────────────────────────────────────
    print(f"\n{separador}")
    print(BOLD + MAGENTA + "       POLIMORFISMO — método describir()" + RESET)
    print(separador)

    figuras: list[FigurasGeometricas] = [
        FigurasGeometricas(3, 7),
        Rectangulo(8, 15),
        Triangulo(6, 5),
        Elipse(10, 4),
    ]

    print("\n  Recorriendo la lista con un mismo llamado → figura.describir()\n")
    for i, figura in enumerate(figuras, start=1):
        color = figura.__class__.COLOR if hasattr(figura.__class__, "COLOR") \
                else "\033[97m"
        print(f"  [{i}] {color}{figura.describir()}{RESET}")
        print(f"       Área: {figura.area():.2f} cm²  |  "
              f"Perímetro: {figura.perimetro():.2f} cm\n")

    print(separador)
    print(BOLD + CYAN + "  Código ejecutado correctamente   " + RESET)
    print(separador)
```
